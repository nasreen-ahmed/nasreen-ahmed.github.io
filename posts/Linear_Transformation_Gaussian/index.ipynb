# Linear Transformation of a Gaussian Distribution

[Theorem : A Linear Transformation of a Gaussian is also a Gaussian]{style="color:RoyalBlue"}

**Proof**: 


Let $\epsilon \sim \mathcal{N}(\mu,\sigma^2)$, so the PDF is 

$$
p(\epsilon) = \frac{1}{\sqrt{2\pi \sigma^2}}\exp (-\frac{(\epsilon-\mu)^2}{2\sigma^2})
$$

Define $t = a+b\epsilon$, where $a,b$ are constants. We want to find $p(t)$

**Step 1** : Find $\epsilon$ in terms of $t$

$$
\epsilon = \frac{t-a}{b}
$$

**Step 2** : Apply the change of variables formula 

For a transformation $t=g(\epsilon)$, the PDf for $t$ is 

$$
p(t) = p(\epsilon) \left|\frac{d\epsilon}{dt}\right|
$$

Compute the Jacobian 

$$
\frac{d\epsilon}{dt} = \frac{1}{b}=> \left|\frac{d\epsilon}{dt}\right| = \frac{1}{|b|}
$$

**Step 3** Substitute

$$
p(t) = \frac{1}{\sqrt{2\pi \sigma^2}}\exp (- \frac{(\frac{t-a}{b}-\mu)^2}{2\sigma^2})\cdot \frac{1}{|b|}
$$



Simplifying the exponent 

$$
\frac{t-a}{b} - \mu = \frac{t-a-b\mu}{b}
$$

So, the squared terrm becomes 

$$
(\frac{t-a-\mu}{b})^2 = \frac{(t-(a+b\mu)^2)}{b^2}
$$

Substituting back 

$$
p(t) = \frac{1}{|b|\sqrt{2\pi \sigma^2}}\exp(- \frac{(t-(a+b\mu))^2}{2b^2\sigma^2})
$$

Rewrite $|b|\sqrt{\sigma^2} = \sqrt{b^2\sigma^2}$

$$
p(t) = \frac{1}{\sqrt{2\pi \sigma^2b^2}}\exp(- \frac{(t-(a+b\mu))^2}{2b^2\sigma^2})
$$

**Step 4** : Recognizing the result 

[This is exactly the Gaussian PDF]{style="color:RoyalBlue"}

$$
\text{mean} = a+b\mu , \text{ variance} = b^2\sigma^2 
$$

[**Therefore,**]{style="color:RoyalBlue"} 

$$
t = a+b\epsilon \sim \mathcal{N}(a+b\mu,b^2\sigma^2)
$$
