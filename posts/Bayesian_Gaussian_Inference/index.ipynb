# [1. Estimating the Mean of a Gaussian Distribution]{style="color : RoyalBlue"}

## [Single Gaussian Random Variable]{style="color : RoyalBlue"}

We shall supposed that the variance $\sigma^2$ is known and we consider the task of inferrng the mean $\mu$ given a set of $\mathbf{N}$ observations.  $X = \{x_1,x_2,\ldots,x_n\}$.

The likelihood function, that is the probability of the observed data given $\mu$ , viewed as a function of $\mu$, is given by : 

$$
p(X|\mu) = \prod_{i=1}^N p(x_n|\mu) = \frac{1}{(2\pi \sigma^2)^{N/2}} \exp
\bigg\{-\frac{1}{2\sigma^2} \sum_{i=1}^N (x_n - \mu)^2\bigg\}
$$

The likelihood function takes the form of the exponential of a quadratic form in $\mu$.

Thus, we choose a prior $p(\mu)$ given by Gaussian, it will be conjugate distribution for this likelihood function because the corresponding posterior will be a product of two exponentials of quadratic functions of $\mu$ and hence will also be Gaussian. 

We therefore, take our prior to be 

$$
p(\mu) = \mathcal{N}(\mu|\mu_0,\sigma_0^2)
$$
and the posterior distribution is given by 

$$
p(\mu|X)=p(X|\mu)p(\mu)
$$

We can show the posterior distribution as : 

$$
p(\mu|X)= \mathcal{N}(\mu|\mu_N,\sigma_N^2) 
$$

where 
$$
\begin{aligned}
\mu_N &= \frac{\sigma^2}{N\sigma_0^2+\sigma^2}\mu_0 + \frac{N\sigma_0^2}{N\sigma_0^2+\sigma^2}\mu_{ML}\\
\frac{1}{\sigma_N^2} &= \frac{1}{\sigma_0^2} +\frac{N}{\sigma^2}
\end{aligned}
\tag{1}
$$




### [Deriving equation in (1)]{style="color:RoyalBlue"}

#### [The algebra]{style="color:RoyalBlue"}

We have $N$  observations $\mathcal{X}=\{x_1,\ldots,x_n\}$ from a Gaussian with known variance $\sigma^2$ and unknown mean $\mu$. We want to infer $\mu$

By Bayes' Theorem 

$$
p(\mu|\mathcal{X}) \propto p(\mathcal{X}|\mu)p(\mu)
$$

Taking logs, we need the exponent of the the product. Ignoring constants 

$$
\log p(\mu|\mathcal{X}) \propto -\frac{1}{2\sigma^2} \sum_{n=1}^N (x_n-\mu)^2 -\frac{1}{2\sigma_0^2}(\mu-\mu_0)^2
$$

[Expand the first sum]{style="color:RoyalBlue"} 

$$
\sum_{n=1}^N (x_n - \mu)^2 = -\frac{1}{2\sigma^2}(\sum_{n=1}^N x_n^2 - 2\mu \sum_n x_n + \sum_n \mu^2)
$$

$\frac{1}{\sigma^2}\sum_n x_n = \frac{N}{\sigma^2}\frac{\sum_n x_n}{N} = \frac{N}{\sigma^2}\mu_{ML}$, since $\mu_{ML}=\frac{1}{N}\sum_n x_n$

Sorting by Power of $\mu$

$$
\begin{array}{l|c|c}
\text{Term} & \text{Power of} \mu^* & \text{Keep }  \\
\hline
- \frac{1}{2\sigma^2} \sum_{n=1}^N x_n^2 & \mu^0 \text{ constant } & \text { No  -absorbed into normalization} \\
\hline
\frac{\mu}{\sigma^2} \sum_{n=1}^N x_n & \mu^1 & \text { Yes}  \\
\hline
- \frac{N}{2\sigma^2} \mu^2 & \mu^2 & \text {Yes}  \\
\end{array}
$$

So, the likelihood contributes:

$$
-\frac{N}{2\sigma^2}\mu^2 +\frac{N\mu_{ML}}{\sigma^2}\mu 
$$


[Expand the second sum ( Prior term)]{style="color:RoyalBlue"} 

$$
-\frac{1}{2\sigma_0^2}(\mu^2 - 2\mu_0 \mu +\mu_0^2)
$$

Sorting by the power of $\mu$
$$
\begin{array}{l|c|c}
\text{Term} & \text{Power of} \mu^* & \text{Keep }  \\
\hline
-\frac{1}{2\sigma_0^2}\mu_0^2 & \mu^0 \text( constant) & {No}\\
\hline
\frac{\mu_0}{\sigma_0^2}\mu & \mu^1 & {Yes}\\
\hline
- \frac{1}{2\sigma_0^2} \mu^2& \mu^2 & {Yes}
\end{array}
$$

$$
- \frac{1}{2\sigma_0^2} \mu^2 + \frac{\mu_0}{\sigma_0^2}\mu 
$$



$ The \mu^2$ coefficients add : $-\frac{N}{2\sigma^2}\mu^2 -\frac{1}{2\sigma_0^2}\mu^2 = - \frac{1}{2}(\frac{N}{\sigma^2} + \frac{1}{\sigma_0^2})\mu^2$

The $\mu^1$ coefficient : $\frac{N\mu_{ML}}{\sigma^2}\mu + \frac{\mu_0}{\sigma_0^2}\mu$ 

Which gives us the expressions

$$
\propto -\frac{1}{2}(\frac{N}{\sigma^2}+\frac{1}{\sigma_0^2})\mu^2 + (\frac{N\mu_{ML}}{\sigma^2}+\frac{\mu_0}{\sigma_0^2})\mu
$$

Call the above expression : $-\frac{A}{2}\mu^2 + B\mu$

$$
A = \frac{N}{\sigma^2} + \frac{1}{\sigma_0^2} , B = \frac{N\mu_{ML}}{\sigma^2} + \frac{\mu_0}{\sigma_0^2}
$$

Factor - $\frac{A}{2}$ out of both terms : 

$$
-\frac{A}{2}\mu^2 + B\mu = -\frac{A}{2}(\mu^2 + \frac{2B}{A}\mu)
$$

Add and substract $(B/A)^2$ inside the bracket to make it a perfect square 

$$
\begin{aligned}
&-\frac{A}{2}(\mu^2 + \frac{2B}{A}\mu+\frac{B^2}{A^2}-\frac{A^2}{B^2})\\
& -\frac{A}{2}(\mu - \frac{B}{A})^2 + \underbrace{\frac{B^2}{2A}}_{\text{constant in }\mu}
\end{aligned}
$$

The last term has no $\mu$ - it's absorbed into the normalizing constant. What remains is 

$$
\propto - \frac{A}{2}(\mu - \frac{B}{A})^2
$$

[That is exactly $\log \mathcal{N}(\mu | \frac{B}{A},\frac{A}{2})$]{style="color:RoyalBlue"}

$$
\sigma_N^2 = \frac{1}{A}, \text{ }, \mu_N = \frac{B}{A}
$$

[Compute $\sigma_N^2 =1/A$]{style="color:RoyalBlue"}

$$
\begin{aligned}
\frac{1}{\sigma_N^2} = A &= \frac{N}{\sigma^2} + \frac{1}{\sigma_0^2}\\
&= \frac{N\sigma_0^2+\sigma^2}{\sigma^2\sigma_0^2}\\
\end{aligned}
$$

So:

$$
\sigma_N^2 = \frac{\sigma^2\sigma_0^2}{N\sigma_0^2+\sigma^2}
$$


[Compute $\mu_n= B/A$]{style="color:RoyalBlue"}

$$
\mu_n = \frac{B}{A} = B\cdot\frac{1}{A} = B\cdot \sigma_N^2
$$

Substituting $B$ and $\sigma_n^2$

$$
\mu_n = (\frac{N\mu_{ML}}{\sigma^2}+\frac{\mu_0}{\sigma_0^2}) \cdot \frac{\sigma^2\sigma_0^2}{N\sigma_0^2+\sigma^2}
$$

First term 

$$
\frac{N\mu_{ML}}{\sigma^2} \cdot \frac{\sigma^2\sigma_0^2}{N\sigma_0^2+\sigma^2}
$$

$$
(\frac{N\sigma_0^2}{N\sigma_0^2+\sigma^2})\mu_{ML}
$$

Second Term

$$
\frac{\sigma^2}{N\sigma_0^2+\sigma^2}\mu_0
$$

Adding first and second term 

$$
\mu_N = \frac{\sigma^2}{N\sigma_0^2+\sigma^2}\mu_0 + \frac{N\sigma_0^2}{N\sigma_0^2+\sigma^2}\mu_{ML}
$$

The two weights $\frac{\sigma^2}{N\sigma_0^2+\sigma^2}$ and $\frac{N\sigma_0^2}{N\sigma_0^2+\sigma^2}$ visibly sum to 1, confirming $\mu_N$ is a convex combination of prior mean and MLE. 

[ To derive how the Gaussian $\mathcal{N}(\mu|m,v)\propto -\frac{1}{2v}\mu^2+\frac{m}{v}\mu$]{style="color:RoyalBlue"}

The Gaussian $\mathcal{N}(\mu | m,v)$ is defined as 

$$
\mathcal{N}(\mu|m,v) = \frac{1}{\sqrt{2\pi v}}\exp(-\frac{(\mu-m)^2}{2v})
$$

Take the log both sides 

$$
\mathcal{N}(\mu|m,v) = \log \frac{1}{\sqrt{2\pi v}} - \frac{(\mu-m)^2}{2v}
$$

The first term, $\log \frac{1}{\sqrt{2\pi v}}$, contains no $mu$ at all - it is a pure constant. We drop it 

$$
\mathcal{N}(\mu|m,v) \propto  - \frac{1}{2v}(\mu-m)^2
$$

Expand the square $(\mu-m)^2 = \mu^2 - 2\mu m +m^2$

$$
\begin{aligned}
&\propto -\frac{1}{2v}(\mu^2 - 2\mu m +m^2)\\
&\propto -\frac{1}{2v}\mu^2 +\frac{m}{v}\mu -\frac{1}{2v} m^2
\end{aligned}
$$

The last term $-\frac{1}{2v}m^2$ contains no $\mu$ - it is another constant. So dropping it. 

$$
\log \mathcal{N}(\mu|m,v) \propto -\frac{1}{2v}\mu^2 +\frac{m}{v}\mu
$$

[This is the key result. Every Gaussian, no matter what its mean $m$ and variance $v$ are, always produces exactly this pattern when we take it log and keep only the $\mu$ dependent terms]{style="color:RoyalBlue"}

Now going back to the log posterior of the Gaussian 

$$
\log p(\mu|\mathcal{X}) \propto -\frac{1}{2}(\frac{N}{\sigma^2} + \frac{1}{\sigma_0^2})\mu^2 +(\frac{N\mu_{ML}}{\sigma^2}+\frac{\mu_0}{\sigma_0^2})\mu
$$

A Gaussian $\mathcal{N}(\mu|m,v)$ in log form looks like:

$$
\log \mathcal{N}(\mu|m,v) \propto -\frac{1}{2v}\mu^2 +\frac{m}{v}\mu
$$

So, the coefficient of $\mu^2$ in any Gaussian log density is always $-\frac{1}{2v}$. Matching that we have 

$$
\begin{aligned}
&-\frac{1}{2\sigma_N^2} = -\frac{1}{2}(\frac{N}{\sigma^2} +\frac{1}{\sigma_0^2})\\
&\frac{1}{\sigma_N^2} = \frac{N}{\sigma^2} + \frac{1}{\sigma_0^2}
\end{aligned}
$$

[**Therefore, we get**]{style="color:RoyalBlue"}

$$
\begin{aligned}
&\mu_N = \frac{\sigma^2}{N\sigma_0^2+\sigma^2} \mu_0 + \frac{N\sigma_0^2}{N\sigma_0^2+\sigma^2} \mu_{ML}\\
&\frac{1}{\sigma_N^2} = \frac{1}{\sigma_0^2} + \frac{N}{\sigma^2}
\end{aligned}
$$

in which $\mu_{ML}$ is the maximum likelihood for $\mu$ given by the sample mean 

$$
\mu_{ML} = \frac{1}{N}\sum_{n=1}^N x_n
$$

- The mean of the posterior distribution comprise between the prior mean $\mu_0$ and the maximum likelihood solution $\mu_{ML}$
- If the number of observed data points $N=0$ , then the mean equation above reduces to the prior mean as expected. 
- For $N \rightarrow \infty$ , the posterior mean is given by the maximum likelihood solution.
- The variance is given in terms of precision . The precisions are additive, so that the precision of the posterior is given by the precision of the prior plus one contributionof the data precision from each of the observed data points.
- As we increase the number of data points , the precision steadily increases, corresponding to a posterior distribution with steadily decreasing variance.

We therefore, see that the maximum likelihood result of a point estimate for $\mu$ is recovered precisely from the Bayesian formalism in the limit of an infinite number of observations.

# [2. Estimating the Variance of a Gaussian Distribution]{style="color : RoyalBlue"}

Now suppose mean is known and we wish to infer variance. 
It is more convenient to work with the precision . $\lambda = 1/\sigma^2$. The likelihood function for $\lambda$ takes the form 

$$
p(X|\lambda) = \prod_{n=1}^N \mathcal{N}(x_n|\mu,\lambda^{-1}) \propto \lambda^{N/2}\exp \{ \frac{\lambda}{2}\sum_{n=1}^N (x_n-\mu)^2 \}
$$
